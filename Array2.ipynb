#Reverse Rotate

```python
arr=[10,20,30,40,50]
left=0
right=len(arr)-1

while left <right:
  arr[left],arr[right]=arr[right],arr[left]
  left+=1
  right-=1
print(arr)

```

Liner Search

```python
arr = [12, 45, 23, 67, 34]
target = 23
for i in range(len(arr)):
  if arr[i]== target:
    print(f"target found at index {i}")
    break
else:
  print("target not found")
```

Bubble Sorting

```python
arr=[2,4,14,2,5,1,52,90]
for i in range(len(arr)):
  for j in range(i+1,len(arr)):
    if arr[i]>arr[j]:
      arr[i],arr[j]=arr[j],arr[i]

print(arr)
```

Selection Sort

```python
arr=[34,2,5,1,66,53]
for i in range(len(arr)):
  min_index=i
  for j in range(i+1,len(arr)):
    if arr[j]<arr[min_index]:
      min_index=j
      arr[i],arr[min_index]=arr[min_index],arr[i]
print(arr)
```

Insertion Sort

```python
arr = [5, 2, 8, 1, 3]

for i in range(1, len(arr)):
    key = arr[i]
    j = i - 1

    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1

    arr[j + 1] = key

print(arr)
```
