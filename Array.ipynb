```python
arr = [1,2,3,5,6]

def  missing_value (arr):
  n=len(arr)+1

  xor=0



  for i in range(1,n+1):

    xor=xor^i

  for i in arr:

    xor=xor^i

  return xor
print(missing_value(arr))
```

```python
arr = [1, 2, 3, 5]

def missing_number(arr):

    n = len(arr) + 1

    xor = 0

    for i in range(1, n + 1):
        xor = xor ^ i

    for i in arr:
        xor = xor ^ i

    return xor

print(missing_number(arr))
```

```python
arr=[1,2,3,4,6,7,9,11,13]

def missing_value(arr):
  n=len(arr)+1
  xor=0
  for i in range(1,n+1):
    xor=xor^i
  for i in arr:
    xor=xor^i
    return xor
print(missing_value(arr))
```

```python
arr=[1,3,5,7,8,9,11]
def missing_multiple (arr,n) :
  missing_num=[]
  for i in range(1,n+1):
    if i  not in arr:
      missing_num.append(i)
  return missing_num

print(missing_multiple(arr,11))

```
